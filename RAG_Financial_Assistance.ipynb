```python
import os

def load_documents(folder="documents"):
    documents = []

    for filename in os.listdir(folder):
        filepath = os.path.join(folder, filename)

        if filename.endswith(".txt"):
            with open(filepath, "r", encoding="utf-8") as f:
                text = f.read()

            documents.append({
                "source": filename,
                "text": text
            })

    return documents



documents = load_documents()

for doc in documents:
    print(doc["source"])
```

```python
def split_into_chunks(text):
    # Split document into separate sections using blank lines
    sections = text.split("\n\n")

    chunks = []

    for section in sections:
        section = section.strip()

        if section:
            chunks.append(section)

    return chunks




all_chunks = []

for doc in documents:

    chunks = split_into_chunks(doc["text"])

    for chunk in chunks:

        all_chunks.append({
            "source": doc["source"],
            "text": chunk
        })


print("Total chunks:", len(all_chunks))

for i, chunk in enumerate(all_chunks):

    print("\n" + "=" * 70)
    print(f"CHUNK {i + 1}")
    print(f"Source: {chunk['source']}")
    print("=" * 70)

    print(chunk["text"])



```

```python
#loading embedding model

from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
texts = [chunk["text"] for chunk in all_chunks]

embeddings = model.encode(
    texts,
    convert_to_numpy=True
)

print(embeddings.shape)
```

```python
#creating FAISS Vector Database

import faiss
import numpy as np

dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(np.array(embeddings).astype("float32"))
```

```python
#retrieval

def retrieve_documents(query, company=None, top_k=3):

    # ------------------------------------------------
    # 1. Select documents for the requested company
    # ------------------------------------------------

    if company is not None:
        company_chunks = []

        for chunk in all_chunks:
            if company.lower() in chunk["source"].lower():
                company_chunks.append(chunk)
    else:
        company_chunks = all_chunks

    if len(company_chunks) == 0:
        print("No documents found for:", company)
        return []


    # ------------------------------------------------
    # 2. Convert the query into an embedding
    # ------------------------------------------------

    query_embedding = model.encode(
        [query],
        convert_to_numpy=True
    ).astype("float32")


    # ------------------------------------------------
    # 3. Create embeddings for the selected chunks
    # ------------------------------------------------

    chunk_texts = [
        chunk["text"] for chunk in company_chunks
    ]

    chunk_embeddings = model.encode(
        chunk_texts,
        convert_to_numpy=True
    ).astype("float32")


    # ------------------------------------------------
    # 4. Create FAISS index
    # ------------------------------------------------

    dimension = chunk_embeddings.shape[1]

    temp_index = faiss.IndexFlatL2(dimension)

    temp_index.add(chunk_embeddings)


    # ------------------------------------------------
    # 5. Search FAISS
    # ------------------------------------------------

    k = min(top_k, len(company_chunks))

    distances, indices = temp_index.search(
        query_embedding,
        k
    )


    # ------------------------------------------------
    # 6. Collect the retrieved evidence
    # ------------------------------------------------

    results = []

    for i, idx in enumerate(indices[0]):

        results.append({
            "source": company_chunks[idx]["source"],
            "evidence": company_chunks[idx]["text"],
            "distance": float(distances[0][i])
        })

    return results


```

```python
import os
import faiss
import numpy as np
from sentence_transformers import SentenceTransformer

class FinancialRAG:

    def __init__(self, documents_folder="documents"):

        print("Loading embedding model...")

        self.model = SentenceTransformer(
            "all-MiniLM-L6-v2"
        )

        self.documents_folder = documents_folder

        self.chunks = []

        # Load and prepare documents
        self.load_documents()

        # Create embeddings and FAISS index
        self.build_index()

        print("RAG system ready!")


    # --------------------------------------------------
    # 1. LOAD DOCUMENTS
    # --------------------------------------------------

    def load_documents(self):

        print("\nLoading documents...")

        for filename in os.listdir(
            self.documents_folder
        ):

            if not filename.endswith(".txt"):
                continue


            filepath = os.path.join(
                self.documents_folder,
                filename
            )


            with open(
                filepath,
                "r",
                encoding="utf-8"
            ) as f:

                text = f.read()


            # Identify company from filename
            company_name = filename.replace(
                ".txt",
                ""
            ).replace(
                "_",
                " "
            ).title()


            # Split into sections
            sections = text.split("\n\n")


            current_section = "General"


            for section in sections:

                section = section.strip()

                if not section:
                    continue


                # Determine section name
                first_line = section.split(
                    "\n"
                )[0].strip()


                if first_line.startswith(
                    "Company:"
                ):

                    current_section = "Company"


                elif first_line.startswith(
                    "Financial Performance:"
                ):

                    current_section = (
                        "Financial Performance"
                    )


                elif first_line.startswith(
                    "Growth Factors:"
                ):

                    current_section = (
                        "Growth Factors"
                    )


                elif first_line.startswith(
                    "Risks:"
                ):

                    current_section = "Risks"


                self.chunks.append({

                    "company": company_name,

                    "source": filename,

                    "section": current_section,

                    "text": section

                })


        print(
            "Total chunks:",
            len(self.chunks)
        )

    # --------------------------------------------------
    # 2. BUILD FAISS INDEX
    # --------------------------------------------------

    def build_index(self):

        print("\nCreating embeddings...")

        texts = [
            chunk["text"]
            for chunk in self.chunks
        ]

        embeddings = self.model.encode(
            texts,
            convert_to_numpy=True
        ).astype("float32")


        dimension = embeddings.shape[1]


        # Create FAISS index
        self.index = faiss.IndexFlatL2(
            dimension
        )


        # Add embeddings
        self.index.add(embeddings)


        print("FAISS index created.")
        print(
            "Number of vectors:",
            self.index.ntotal
        )


    # --------------------------------------------------
    # 3. RETRIEVE DOCUMENTS
    # --------------------------------------------------

    def retrieve(
        self,
        company,
        query,
        top_k=3
    ):

        # Find chunks belonging to the company
        company_chunks = []

        for chunk in self.chunks:

            source = chunk["source"].lower()

            if company.lower() in source:

                company_chunks.append(chunk)


        # If no company found
        if not company_chunks:

            return []


        # Convert query into embedding
        query_embedding = self.model.encode(
            [query],
            convert_to_numpy=True
        ).astype("float32")


        # Get embeddings for company's chunks
        chunk_texts = [
            chunk["text"]
            for chunk in company_chunks
        ]

        chunk_embeddings = self.model.encode(
            chunk_texts,
            convert_to_numpy=True
        ).astype("float32")


        # Create temporary FAISS index
        dimension = chunk_embeddings.shape[1]

        company_index = faiss.IndexFlatL2(
            dimension
        )

        company_index.add(chunk_embeddings)


        # Number of results cannot exceed
        # number of available chunks
        k = min(
            top_k,
            len(company_chunks)
        )


        # Search
        distances, indices = company_index.search(
            query_embedding,
            k
        )


        # Prepare results
        results = []


        for i, idx in enumerate(indices[0]):

            distance = float(
                distances[0][i]
            )


            results.append({

                "source":
                    company_chunks[idx]["source"],

                "evidence":
                    company_chunks[idx]["text"],

                "distance":
                    distance
            })


        return results


    # --------------------------------------------------
    # 4. RELEVANCE LABEL
    # --------------------------------------------------

    def relevance_label(self, distance):

        if distance < 0.5:

            return "Highly Relevant"

        elif distance < 1.0:

            return "Relevant"

        else:

            return "Less Relevant"


    # --------------------------------------------------
    # 5. RETRIEVE + ADD RELEVANCE
    # --------------------------------------------------

    def retrieve_with_relevance(
        self,
        company,
        query,
        top_k=3
    ):

        results = self.retrieve(
            company,
            query,
            top_k
        )


        for result in results:

            result["relevance"] = (
                self.relevance_label(
                    result["distance"]
                )
            )


        return results


    def format_evidence(self, results):

      if not results:

        return "No relevant evidence was found."


      formatted = ""


      for i, result in enumerate(results):

        formatted += (
            f"\nEvidence {i + 1}\n"
        )

        formatted += (
            f"Source: {result['source']}\n"
        )

        formatted += (
            f"Relevance: {result['relevance']}\n"
        )

        formatted += (
            f"Text: {result['evidence']}\n"
        )

        formatted += (
            "-" * 50
            + "\n"
        )


      return formatted
```
